```python
import numpy as np
import pandas as pd
import seaborn as sns  
```

```python
df = sns.load_dataset("diamonds")
df.head()
```

```python
## checking null values
df.isnull().sum()
```

```python
## separating data as x and y
y = df[["price"]]
x = df.drop(labels="price", axis=1)
```

```python
x.info()
```

```python
x_num = x.select_dtypes(include="float64")
x_num.head()
```

```python
x_char = x.select_dtypes(exclude="float64")
x_char.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
for col in x_char:
    x_char[col] = le.fit_transform(x_char[col])
x_char.head()
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
x_scaling = pd.DataFrame(sc.fit_transform(x_num), 
                         columns=x_num.columns,
                         index=x_num.index)
x_scaling.head()
```

```python
data = pd.concat([x_char , x_scaling,], axis=1)
data.head() 
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(data, y, 
                                                    test_size=0.2, 
                                                    random_state=45)
```
