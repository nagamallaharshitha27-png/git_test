```python
df.drop_duplicates(inplace=True)
```

```python
## ml - sklearn
y= df['Car_Price_INR']
x = df.drop(labels="Car_Price_INR",axis=1)
```

```python
x_num = x.select_dtypes(include="int64")
x_num.head()
```

```python
x_char = x.select_dtypes(exclude="int64")
x_char.head()
```

```python
## char
## nominal -> one hot enocoder
## ordinal ->label encoder
```

```python
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)
```

```python
# fit-learning transform - applying
## fit-transform - learning and applying
encoded_data = ohe.fit_transform(x_char)
x_encoded = pd.DataFrame(encoded_data,
                         columns=ohe.get_feature_names_out(x_char.columns),
                         index = x_char.index)
x_encoded.head()
```

```python
# num - scaling
## feature1 - 1 -10 , feature2 - 1-100000
## standard scalar : -3 to 3
## min max scaling : 0 to 1
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
x_scaling = pd.DataFrame(sc.fit_transform(x_num),
                       columns=x_num.columns,
                       index=x_num.index)
x_scaling.head()
```

```python
data= pd.concat([x_scaling,x_encoded],axis=1)
data.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(data,y,test_size=0.2,random_state=56)
```

```python
print(df.shape)
print(x_train.shape)
print(x_test.shape)
```

```python
y_train
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
ypred = lr.predict(x_test)
ypred
```
